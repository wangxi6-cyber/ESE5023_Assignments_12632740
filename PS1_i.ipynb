# 1. Flowchart

```python
#1
def Print_values(a, b, c):

    if a > b:
        if b > c:
            print(a, b, c)
        else:
            if a > c:
                print(a, c, b)
            else:
                print(c, a, b)
    else:
        if b > c:
            if a > c:
                print(b, a, c)
            else:
                print(b, c, a)
        else:
            print(c, b, a)


# Random test
Print_values(9, 5, 2)
Print_values(8, 3, 6)
Print_values(2, 5, 8)
Print_values(5, 9, 3)
```

# 2. Matrix multiplication

```python
#2
import random


# =========================
# 2.1 Create two matrices
# =========================

# M1: 5 rows and 10 columns
M1 = [
    [random.randint(0, 50) for j in range(10)]
    for i in range(5)
]


# M2: 10 rows and 5 columns
M2 = [
    [random.randint(0, 50) for j in range(5)]
    for i in range(10)
]


print("M1:")
for row in M1:
    print(row)


print("\nM2:")
for row in M2:
    print(row)



# =========================
# 2.2 Matrix multiplication
# =========================

def Matrix_multip(M1, M2):

    # M1 size: 5 x 10
    # M2 size: 10 x 5

    rows_M1 = len(M1)
    cols_M1 = len(M1[0])

    cols_M2 = len(M2[0])


    # create result matrix (5 x 5)
    result = [
        [0 for j in range(cols_M2)]
        for i in range(rows_M1)
    ]


    # matrix multiplication
    for i in range(rows_M1):          # rows of M1
        for j in range(cols_M2):      # columns of M2
            for k in range(cols_M1):  # common dimension

                result[i][j] = result[i][j] + M1[i][k] * M2[k][j]


    return result



# Calculate M1 * M2
M3 = Matrix_multip(M1, M2)


print("\nM1 * M2:")
for row in M3:
    print(row)
```

# 3. Pascal triangle

```python
#3
def Pascal_triangle(k):

    # start from row 0
    row = [1]

    # generate rows until kth row
    for i in range(k):

        new_row = [1]

        # calculate middle numbers
        for j in range(len(row)-1):
            new_row.append(row[j] + row[j+1])

        # last number is always 1
        new_row.append(1)

        row = new_row

    print(row)


# Test
Pascal_triangle(5)

Pascal_triangle(100)

Pascal_triangle(200)
```

# 4. Add or double

```python
#4
def Least_moves(x):

    # base case
    if x == 1:
        return 0


    # if x is even
    if x % 2 == 0:
        return 1 + min(Least_moves(x-1),
                       Least_moves(x//2))

    # if x is odd
    else:
        return 1 + Least_moves(x-1)



print(Least_moves(2))
print(Least_moves(5))
print(Least_moves(100))
```

# 5. Dynamic programming

```python
import matplotlib.pyplot as plt


# 生成所有可能表达式
def generate_expressions():
    digits = "123456789"
    results = []

    def dfs(index, expression, value):
        # 到达最后一个数字
        if index == len(digits):
            results.append((expression, value))
            return

        # 从当前位置开始连接数字
        number = 0
        for i in range(index, len(digits)):
            number = number * 10 + int(digits[i])

            # 第一部分不能有前导0（这里数字没有0，所以不用处理）
            if index == 0:
                # 第一个数字不能带符号
                dfs(i + 1, str(number), number)
            else:
                # 加法
                dfs(i + 1,
                    expression + " + " + str(number),
                    value + number)

                # 减法
                dfs(i + 1,
                    expression + " - " + str(number),
                    value - number)

    dfs(0, "", 0)

    return results


# 预先计算所有结果
all_expressions = generate_expressions()


# 5.1
def Find_expression(target):
    count = 0

    for exp, value in all_expressions:
        if value == target:
            print(exp, "=", target)
            count += 1

    print("Total solutions:", count)


# 5.2
Total_solutions = []

for i in range(1, 101):
    count = 0

    for exp, value in all_expressions:
        if value == i:
            count += 1

    Total_solutions.append(count)


# 最大和最小
max_value = max(Total_solutions)
min_value = min(Total_solutions)

max_numbers = [
    i + 1 for i, x in enumerate(Total_solutions)
    if x == max_value
]

min_numbers = [
    i + 1 for i, x in enumerate(Total_solutions)
    if x == min_value
]


print("Maximum number of solutions:")
print(max_numbers, "with", max_value, "solutions")

print("Minimum number of solutions:")
print(min_numbers, "with", min_value, "solutions")


# Plot
plt.figure(figsize=(10, 5))

plt.plot(range(1, 101),
         Total_solutions,
         marker='o')

plt.xlabel("Target number")
plt.ylabel("Number of solutions")
plt.title("Number of expressions for each number (1-100)")

plt.grid(True)

plt.show()


# 测试题目例子
Find_expression(50)
Find_expression(88)
```
